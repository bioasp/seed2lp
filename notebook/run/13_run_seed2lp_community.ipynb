### Execute

```python
from os import path, listdir
```

```python
bigg_sbml_dir = f"{data_dir}/BiGG"
toy_sbml_dir = f"{data_dir}/toys"
com_description_dir = f"{data_dir}/descriptions"
bigg_1st_result_dir = f"{result_dir}/com_bigg_1st_results"
bigg_1min_result_dir = f"{result_dir}/com_bigg_1min_results"
toys_result_dir = f"{result_dir}/com_toys"


heuristic=["global", "bisteps", "delsupset"]
```

This function will execute seed2lp community for toy:
- *Global*, *Bistep*, *Delete superset*
- no accumulation
- Limitations: 1 min

Also, it will check the flux for each solution and write it into files.

```python
def run_s2lp(result_dir:str,is_toy:bool=False,is_first:bool=False): 
    supp_first = ''
    if is_first:
        supp_first = '-nbs 1'
    if is_toy:
        sbml_dir = toy_sbml_dir
    else:
        sbml_dir = bigg_sbml_dir
    for filename in listdir(com_description_dir):
        if is_toy and "com_toy" in filename \
            or not is_toy and "com_toy" not in filename:
                for heur in heuristic:
                    desc_path = path.join(com_description_dir, filename)
                    command = f"{desc_path} {sbml_dir} {result_dir} -cm {heur} --temp {temp_dir} -tl {time_limit} {supp_first} -cf"
                    !seed2lp community {command}
```

The execution might take more than 30min due to finding minimal solutions and due to *Hybrid-lpx* mode (requires lot of time to calculate fluxes)

```python
run_s2lp(toys_result_dir, is_toy=True)
```

```python
run_s2lp(bigg_1st_result_dir, is_first=True)
```

```python
run_s2lp(bigg_1min_result_dir)
```
